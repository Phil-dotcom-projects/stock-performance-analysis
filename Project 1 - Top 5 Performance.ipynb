# Stock Performance Analysis: Top 5 Holdings vs S&P 500

## Project Overview
This project shows the performance of my top 5 best perfoming stocks over the last 4 years (May 2022 - May 2026) in comparison of the S&P 500 performance.

## Stocks Analyzed
- **Vertiv Holdings** (VRT)
- **Sterling Infrastructure** (STRL)
- **Howmet Aerospace** (HWM)
- **ACM Research** (ACMR)
- **Badger Infrastructure Solutions** (BDGI.TO)

## Key Findings
All 5 holdings significantly outperformed the S&P 500 (+72.80%) over the period:
- VRT: +2,377.83% | STRL: +2,255.25% | HWM: +653.21% | ACMR: +249.59% | BDGI.TO: +143.65%
## Libraries Used
yFinance, Pandas, Numpy, Seaborn, Matplotlib 

```python
import yfinance as yf
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Downloading the stock price information

hist_data = yf.download(["HWM", "VRT", "STRL", "ACMR",  "BDGI.TO", "^GSPC"], start = "2022-05-01", end = "2026-05-01")
```

```python
# Printing the data to see if it downloaded correctly

hist_data
```

```python
# Selecting only the closing price and displaying the first 10 rows

close_price = hist_data["Close"]
close_price.head(10)
```

```python
# Computing daily returns to start at the same baseline

daily_returns = close_price.pct_change()
daily_returns.head(10)
```

```python
# Computing cumulative returns to see how much $1 invested grew over time

cumulative_returns = (1 + daily_returns).cumprod() - 1
cumulative_returns
```

```python
# Visualizing the cumulative returns of ACMR, BDGI.TO, HWM, STRL, VRT versus the S&P 500

cumulative_returns.plot(figsize = (12, 6))
plt.ylabel("Cumulative Returns")
plt.title("Cumulative Returns of 5 best performing stock versus the S&P 500");
```

```python
# Building a summary table - Extracting the last row of the cumulative return from the dataset 

total_returns = cumulative_returns.iloc[1023]
total_returns
```

```python
# Building a summary table - Getting the annualized return (To adjust the total return to a "per year" basis)

annualized_returns = (1 + total_returns) ** (1 / 4) - 1
annualized_returns
```

```python
# Building a summary table - Computing the standard deviation on an annualized basis

std = daily_returns.std() * np.sqrt(252)
std
```

```python
# Building a summary table - Finalizing the Table

data = {"Total Returns": total_returns,
        "Annualized Returns": annualized_returns,
        "Annual Standard Deviation": std}

summary_table = pd.DataFrame(data, columns = ["Total Returns", "Annualized Returns", "Annual Standard Deviation"])
summary_table = summary_table.sort_values(by = "Total Returns", ascending = False) # Sorting the table to be organized as highest total returns to lowest total returns
summary_table.style.format('{:.2%}') # Applying a uniform percentage format with 2 decimal places to all numeric columns in the DataFrame "summary_table"
```
